```python
import re
import json
import random
from datetime import datetime
from pathlib import Path


# ============================================================
# SIMPLE RULE-BASED CHATBOT
# ============================================================

class ChatBot:

    def __init__(self):
        self.name = "Byte"
        self.user_name = "User"
        self.history = []
        self.history_file = Path("chat_log.json")

        # ----------------------------------------------------
        # INTENT DEFINITIONS
        # ----------------------------------------------------
        self.intent_patterns = {

            "greeting": [
                "hello",
                "hi",
                "hey",
                "good morning",
                "good afternoon",
                "good evening"
            ],

            "farewell": [
                "bye",
                "goodbye",
                "see you",
                "exit",
                "quit"
            ],

            "thanks": [
                "thank you",
                "thanks",
                "thank u",
                "thx"
            ],

            "help": [
                "help",
                "what can you do",
                "commands",
                "options"
            ],

            "bot_identity": [
                "who are you",
                "what are you",
                "your name",
                "tell me about yourself"
            ],

            "wellbeing": [
                "how are you",
                "how are things",
                "are you okay",
                "how do you feel"
            ],

            "joke": [
                "tell me a joke",
                "make me laugh",
                "joke"
            ]
        }

        # ----------------------------------------------------
        # KNOWLEDGE BASE
        # ----------------------------------------------------
        self.knowledge = {

            "python": (
                "Python is a high-level programming language known "
                "for its readable syntax. It is commonly used in "
                "software development, automation, data science and AI."
            ),

            "ai": (
                "Artificial Intelligence is the field of computer "
                "science concerned with creating systems capable of "
                "performing tasks that normally require human intelligence."
            ),

            "machine learning": (
                "Machine Learning is a branch of AI where computers "
                "learn patterns from data and use those patterns to "
                "make predictions or decisions."
            ),

            "algorithm": (
                "An algorithm is a sequence of well-defined steps "
                "used to solve a particular problem."
            ),

            "database": (
                "A database is an organized collection of information "
                "that can be stored, retrieved and managed efficiently."
            ),

            "networking": (
                "Computer networking allows computers and other devices "
                "to communicate and exchange information."
            ),

            "oop": (
                "Object-Oriented Programming organizes software around "
                "objects and classes. Its major principles include "
                "encapsulation, inheritance, abstraction and polymorphism."
            ),

            "recursion": (
                "Recursion occurs when a function calls itself to solve "
                "a smaller version of the same problem."
            ),

            "a star": (
                "A* is a pathfinding algorithm that uses both the "
                "actual cost of reaching a node and a heuristic estimate "
                "of the remaining distance."
            ),

            "sql": (
                "SQL stands for Structured Query Language and is used "
                "to communicate with and manage relational databases."
            )
        }

        # ----------------------------------------------------
        # RESPONSE COLLECTIONS
        # ----------------------------------------------------
        self.responses = {

            "greeting": [
                "Hello! How can I assist you?",
                "Hi! Nice to chat with you.",
                "Hey there! What would you like to ask?"
            ],

            "thanks": [
                "You're welcome!",
                "No problem!",
                "Happy to help!"
            ],

            "wellbeing": [
                "I'm doing great and ready to answer your questions.",
                "I'm functioning perfectly. How about you?",
                "I'm good! What can I help you with?"
            ],

            "joke": [
                "Why do programmers prefer dark mode? Because light attracts bugs!",
                "Why did the computer go to the doctor? Because it had a virus!",
                "There are only 10 kinds of people: those who understand binary and those who don't."
            ],

            "farewell": [
                "Goodbye! Have a great day.",
                "See you next time!",
                "Thanks for chatting with me. Goodbye!"
            ],

            "thanks": [
                "You're welcome!",
                "Anytime!",
                "Glad I could help!"
            ]
        }

    # ========================================================
    # TEXT CLEANING
    # ========================================================

    def clean_text(self, text):

        text = text.lower()

        # Keep letters, numbers and mathematical operators
        text = re.sub(r"[^a-zA-Z0-9\s+\-*/().]", "", text)

        text = re.sub(r"\s+", " ", text)

        return text.strip()

    # ========================================================
    # INTENT CLASSIFICATION
    # ========================================================

    def classify_intent(self, message):

        message = self.clean_text(message)

        highest_score = 0
        selected_intent = None

        for intent, patterns in self.intent_patterns.items():

            current_score = 0

            for phrase in patterns:

                if phrase in message:
                    current_score += len(phrase.split())

            if current_score > highest_score:
                highest_score = current_score
                selected_intent = intent

        return selected_intent

    # ========================================================
    # KNOWLEDGE BASE SEARCH
    # ========================================================

    def find_knowledge(self, message):

        message = self.clean_text(message)

        # Exact topic matching
        for topic, answer in self.knowledge.items():

            if topic in message:
                return answer

        # Additional keyword mapping
        aliases = {
            "what is python": "python",
            "python language": "python",
            "what is ai": "ai",
            "artificial intelligence": "ai",
            "what is ml": "machine learning",
            "machine learning": "machine learning",
            "what is an algorithm": "algorithm",
            "what is a database": "database",
            "computer network": "networking",
            "computer networks": "networking",
            "object oriented programming": "oop",
            "object oriented": "oop",
            "a* algorithm": "a star",
            "astar": "a star",
            "structured query language": "sql"
        }

        for keyword, topic in aliases.items():

            if keyword in message:
                return self.knowledge[topic]

        return None

    # ========================================================
    # USER NAME HANDLING
    # ========================================================

    def identify_user(self, message):

        patterns = [
            r"my name is ([a-zA-Z]+)",
            r"i am ([a-zA-Z]+)",
            r"i'm ([a-zA-Z]+)",
            r"call me ([a-zA-Z]+)"
        ]

        for pattern in patterns:

            result = re.search(
                pattern,
                message,
                re.IGNORECASE
            )

            if result:
                self.user_name = result.group(1).capitalize()

                return (
                    f"Nice to meet you, {self.user_name}! "
                    "I'll remember your name during this conversation."
                )

        return None

    # ========================================================
    # DATE AND TIME FUNCTION
    # ========================================================

    def date_time_response(self, message):

        message = self.clean_text(message)

        current = datetime.now()

        if "time" in message:

            return (
                "The current time is "
                + current.strftime("%I:%M %p")
            )

        if "date" in message or "today" in message:

            return (
                "Today's date is "
                + current.strftime("%d %B %Y")
            )

        return None

    # ========================================================
    # BASIC CALCULATOR
    # ========================================================

    def calculate(self, message):

        message = self.clean_text(message)

        if not message.startswith("calculate"):
            return None

        expression = message.replace("calculate", "", 1).strip()

        if not expression:
            return "Please provide a calculation."

        # Only allow safe mathematical characters
        if not re.fullmatch(
            r"[0-9+\-*/(). ]+",
            expression
        ):
            return "I can only calculate basic arithmetic expressions."

        try:

            result = eval(
                expression,
                {"__builtins__": None},
                {}
            )

            return f"The answer is {result}."

        except ZeroDivisionError:
            return "Division by zero is not allowed."

        except Exception:
            return "I couldn't calculate that expression."

    # ========================================================
    # SPECIAL COMMANDS
    # ========================================================

    def special_command(self, message):

        message = self.clean_text(message)

        if message == "history":
            return self.display_history()

        if message == "clear history":

            self.history.clear()

            try:
                self.history_file.unlink(missing_ok=True)
            except OSError:
                pass

            return "Conversation history has been cleared."

        return None

    # ========================================================
    # RESPONSE GENERATOR
    # ========================================================

    def generate_reply(self, message):

        # 1. Check name
        name_reply = self.identify_user(message)

        if name_reply:
            return name_reply

        # 2. Check date/time
        time_reply = self.date_time_response(message)

        if time_reply:
            return time_reply

        # 3. Check calculator
        calculation = self.calculate(message)

        if calculation:
            return calculation

        # 4. Check special commands
        command_reply = self.special_command(message)

        if command_reply:
            return command_reply

        # 5. Classify intent
        intent = self.classify_intent(message)

        if intent == "bot_identity":

            return (
                f"My name is {self.name}. I am a rule-based chatbot "
                "built using Python, pattern matching, predefined "
                "intents and a knowledge base."
            )

        if intent == "help":

            return (
                "I can handle greetings, small talk, basic calculations, "
                "date and time questions, programming questions, "
                "knowledge-base queries and conversation history."
            )

        if intent in self.responses:

            return random.choice(
                self.responses[intent]
            )

        # 6. Search knowledge base
        knowledge_answer = self.find_knowledge(message)

        if knowledge_answer:
            return knowledge_answer

        # 7. Default response
        return random.choice([
            "I'm not sure about that. Try asking me a programming question.",
            "I don't have a rule for that question yet.",
            "Could you rephrase that?",
            "I couldn't find an appropriate answer in my knowledge base."
        ])

    # ========================================================
    # SAVE CONVERSATION
    # ========================================================

    def save_message(self, user_message, bot_message):

        record = {
            "time": datetime.now().strftime(
                "%Y-%m-%d %H:%M:%S"
            ),
            "user": user_message,
            "bot": bot_message
        }

        self.history.append(record)

        try:

            with open(
                self.history_file,
                "w",
                encoding="utf-8"
            ) as file:

                json.dump(
                    self.history,
                    file,
                    indent=4
                )

        except OSError as error:

            print(
                f"Unable to save conversation: {error}"
            )

    # ========================================================
    # DISPLAY HISTORY
    # ========================================================

    def display_history(self):

        if not self.history:

            return "There is no conversation history yet."

        print("\n========== CHAT HISTORY ==========")

        for record in self.history:

            print(
                f"\n[{record['time']}]"
            )

            print(
                f"You: {record['user']}"
            )

            print(
                f"{self.name}: {record['bot']}"
            )

        print("\n==================================")

        return "Conversation history displayed above."

    # ========================================================
    # START CHAT
    # ========================================================

    def run(self):

        print("\n" + "=" * 55)
        print("              BYTE CHATBOT")
        print("=" * 55)

        print(
            "\nHello! I'm Byte, your rule-based chatbot."
        )

        print(
            "Type 'help' to see my capabilities."
        )

        print(
            "Type 'history' to view previous messages."
        )

        print(
            "Type 'exit' or 'bye' to close the chatbot."
        )

        print("-" * 55)

        while True:

            try:

                user_message = input("\nYou: ").strip()

            except (KeyboardInterrupt, EOFError):

                print(
                    f"\n{self.name}: Goodbye!"
                )

                break

            if not user_message:

                print(
                    f"{self.name}: Please enter a message."
                )

                continue

            # Exit check
            cleaned = self.clean_text(user_message)

            if cleaned in [
                "exit",
                "quit",
                "bye",
                "goodbye"
            ]:

                reply = random.choice(
                    self.responses["farewell"]
                )

                print(
                    f"{self.name}: {reply}"
                )

                self.save_message(
                    user_message,
                    reply
                )

                break

            # Generate response
            reply = self.generate_reply(
                user_message
            )

            print(
                f"{self.name}: {reply}"
            )

            # Save conversation
            self.save_message(
                user_message,
                reply
            )


# ============================================================
# MAIN PROGRAM
# ============================================================

if __name__ == "__main__":

    bot = ChatBot()

    bot.run()
```
